```python
import numpy as np
import pandas as pd

data = {
    'Customer_ID': ['1001', '1002', '1003', '1004', '1005', '1006', '1007', '1008', '1009', '1010', '1011', '1012', '1013', '1014', '1015'],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'Age': [28, 45, 63, 34, 52, 23, 41, 59, 31, 48, 36, 60, 25, 50, 42],
    'Tenure_Months': [12, 48, 3, 24, 60, 1, 36, 72, 8, 54, 18, 65, 5, 42, 30],
    'Contract_Type': ['Monthly', 'Two Year', 'Monthly', 'One Year', 'Two Year', 'Monthly', 'One Year', 'Two Year', 'Monthly', 'Two Year', 'One Year', 'Two Year', 'Monthly', 'One Year', 'Monthly'],
    'Payment_Method': ['Electronic Check', 'Bank Transfer', 'Electronic Check', 'Credit Card', 'Bank Transfer', 'Electronic Check', 'Credit Card', 'Bank Transfer', 'Electronic Check', 'Credit Card', 'Bank Transfer', 'Credit Card', 'Electronic Check', 'Bank Transfer', 'Electronic Check'],
    'Monthly_Charges': [65.5, 89.0, np.nan, 75.2, 105.0, 45.0, 80.0, 115.5, np.nan, 95.0, 70.0, 110.0, 50.0, 85.0, 68.0],
    'Total_Charges': [786.0, 4272.0, 150.0, 1804.8, 6300.0, 45.0, 2880.0, 8316.0, 440.0, np.nan, 1260.0, 7150.0, 250.0, 3570.0, 2040.0],
    'Churn': ['No', 'No', 'Yes', 'No', 'No', 'Yes', 'No', 'No', 'Yes', 'No', 'No', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nSummary Statistics:")
display(df.describe())

print("\nNull Values Count:")
print(df.isnull().sum())
```

```python
mean_monthly = df['Monthly_Charges'].mean()
df.fillna({'Monthly_Charges': mean_monthly}, inplace=True)

median_total = df['Total_Charges'].median()
df.fillna({'Total_Charges': median_total}, inplace=True)

print("Null counts after imputation:")
print(df.isnull().sum())
```

```python
filtered_df = df[df['Monthly_Charges'] > 60]
sorted_top_records = filtered_df.sort_values(by='Total_Charges', ascending=False)
print("Top 8 Records (Monthly Charges > 60, sorted by Total Charges):")
display(sorted_top_records.head(8))

print("\nFrequency of Contract Type:")
print(df['Contract_Type'].value_counts())

print("\nFrequency of Payment Method:")
print(df['Payment_Method'].value_counts())

print("\nImplicit Indexing:")
display(df.iloc[0:3, 0:4])

print("\nExplicit Indexing:")
display(df.loc[0:2, ['Customer_ID', 'Age', 'Contract_Type']])

print("\nCompound Condition 1:")
cond1 = df[(df['Age'] > 40) & (df['Churn'] == 'Yes')]
display(cond1[['Customer_ID', 'Age', 'Monthly_Charges', 'Churn']])

print("\nCompound Condition 2:")
cond2 = df[(df['Tenure_Months'] >= 24) & (df['Contract_Type'] == 'Two Year')]
display(cond2[['Customer_ID', 'Tenure_Months', 'Contract_Type']])

print("\nCompound Condition 3:")
cond3 = df[(df['Monthly_Charges'] > 80) | (df['Gender'] == 'Female')]
display(cond3[['Customer_ID', 'Gender', 'Monthly_Charges']].head())
```

```python
max_charge = df['Monthly_Charges'].max()
min_charge = df['Monthly_Charges'].min()
print("Maximum Monthly Charge:", max_charge)
print("Minimum Monthly Charge:", min_charge)

print("\nAggregations by Contract Type:")
grouped_contract = df.groupby('Contract_Type')
display(grouped_contract.agg({'Monthly_Charges': ['mean', 'min', 'max'], 'Tenure_Months': 'mean'}))

print("\nAggregations by Churn:")
grouped_churn = df.groupby('Churn')
display(grouped_churn.agg({'Total_Charges': ['sum', 'mean'], 'Age': 'median'}))

df['Annual_Estimated_Cost'] = df['Monthly_Charges'] * 12

print("\nSelection of Two Year Group:")
two_year_group = grouped_contract.get_group('Two Year')
display(two_year_group[['Customer_ID', 'Monthly_Charges', 'Total_Charges']])

correlation = df['Monthly_Charges'].corr(df['Total_Charges'])
print("\nCorrelation between Monthly and Total Charges:", correlation)

min_val = df['Monthly_Charges'].min()
max_val = df['Monthly_Charges'].max()
df['Monthly_Charges_Norm'] = (df['Monthly_Charges'] - min_val) / (max_val - min_val)

df_services = pd.DataFrame({
    'Customer_ID': ['1001', '1002', '1003', '1004', '1005', '1006', '1007', '1008', '1009', '1010', '1011', '1012', '1013', '1014', '1015'],
    'Internet_Service': ['Fiber', 'DSL', 'DSL', 'Fiber', 'Fiber', 'DSL', 'Fiber', 'Fiber', 'DSL', 'Fiber', 'DSL', 'Fiber', 'DSL', 'Fiber', 'DSL']
})

merged_df = pd.merge(df, df_services, on='Customer_ID')
print("\nMerged Sample:")
display(merged_df[['Customer_ID', 'Contract_Type', 'Internet_Service']].head())
```
