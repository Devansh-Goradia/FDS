```python
import pandas as pd

df = pd.read_csv('/content/irrigation_prediction.csv')

num_cols = [
    'Soil_pH',
    'Soil_Moisture',
    'Temperature_C',
    'Humidity',
    'Rainfall_mm',
]

summary_list = []

for col in num_cols:
    data = df[col].dropna().tolist()
    n = len(data)

    total_sum = 0
    for val in data:
        total_sum += val
    mean = total_sum / n

    sorted_data = sorted(data)
    mid = n // 2
    if n % 2 == 0:
        median = (sorted_data[mid - 1] + sorted_data[mid]) / 2.0
    else:
        median = float(sorted_data[mid])

    freq_dict = {}
    for val in data:
        rounded_val = round(val, 2)
        if rounded_val in freq_dict:
            freq_dict[rounded_val] += 1
        else:
            freq_dict[rounded_val] = 1

    max_freq = 0
    for count in freq_dict.values():
        if count > max_freq:
            max_freq = count

    mode = None
    for k, v in freq_dict.items():
        if v == max_freq:
            mode = k
            break

    rng = max(data) - min(data)

    if n % 2 == 0:
        lower_half = sorted_data[:mid]
        upper_half = sorted_data[mid:]
    else:
        lower_half = sorted_data[:mid]
        upper_half = sorted_data[mid + 1:]

    m1 = len(lower_half) // 2
    if len(lower_half) % 2 == 0:
        q1 = (lower_half[m1 - 1] + lower_half[m1]) / 2.0
    else:
        q1 = float(lower_half[m1])

    m3 = len(upper_half) // 2
    if len(upper_half) % 2 == 0:
        q3 = (upper_half[m3 - 1] + upper_half[m3]) / 2.0
    else:
        q3 = float(upper_half[m3])

    iqr = q3 - q1

    sq_diff_sum = 0
    for x in data:
        sq_diff_sum += (x - mean) ** 2
    variance = sq_diff_sum / (n - 1)
    std_dev = variance ** 0.5

    summary_list.append({
        'Attribute': col,
        'Mean': round(mean, 2),
        'Median': round(median, 2),
        'Mode': mode,
        'Range': round(rng, 2),
        'Q1': round(q1, 2),
        'Q3': round(q3, 2),
        'IQR': round(iqr, 2),
        'Variance': round(variance, 2),
        'Std Dev': round(std_dev, 2),
    })

df_activity1 = pd.DataFrame(summary_list)
print('ACTIVITY 1: MULTI-ATTRIBUTE STATISTICAL ANALYSIS')
display(df_activity1)
```

```python
sample_data = [12, 15, 12, 18, 20, 22, 25, 12, 18, 30]

def compute_mean(data):
    total = 0
    for x in data:
        total += x
    return total / len(data)

def compute_median(data):
    s = sorted(data)
    n = len(s)
    mid = n // 2
    if n % 2 == 0:
        return (s[mid - 1] + s[mid]) / 2.0
    else:
        return float(s[mid])

def compute_mode(data):
    counts = {}
    for x in data:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1

    max_c = 0
    for count in counts.values():
        if count > max_c:
            max_c = count

    for k, v in counts.items():
        if v == max_c:
            return k

def compute_range(data):
    return max(data) - min(data)

def compute_variance(data):
    m = compute_mean(data)
    n = len(data)
    sq_diff_sum = 0
    for x in data:
        sq_diff_sum += (x - m) ** 2
    return sq_diff_sum / (n - 1)

def compute_std_dev(data):
    return compute_variance(data) ** 0.5

def compute_iqr(data):
    s = sorted(data)
    n = len(s)
    mid = n // 2

    if n % 2 == 0:
        lower = s[:mid]
        upper = s[mid:]
    else:
        lower = s[:mid]
        upper = s[mid + 1:]

    q1 = compute_median(lower)
    q3 = compute_median(upper)
    return q3 - q1, q1, q3


mean_val = compute_mean(sample_data)
median_val = compute_median(sample_data)
mode_val = compute_mode(sample_data)
range_val = compute_range(sample_data)
var_val = compute_variance(sample_data)
std_val = compute_std_dev(sample_data)
iqr_val, q1_val, q3_val = compute_iqr(sample_data)

print('ACTIVITY 2: SAMPLE DATA COMPUTATIONS')
print(f'Sample Data: {sample_data}')
print(f'Mean: {mean_val:.2f}')
print(f'Median: {median_val:.2f}')
print(f'Mode: {mode_val}')
print(f'Range: {range_val:.2f}')
print(f'Q1: {q1_val:.2f}')
print(f'Q3: {q3_val:.2f}')
print(f'IQR: {iqr_val:.2f}')
print(f'Variance: {var_val:.2f}')
print(f'Standard Deviation: {std_val:.2f}')
```

```python
import pandas as pd

df = pd.read_csv('/content/irrigation_prediction.csv')
grouped_attribute = df['Soil_Moisture'].dropna().tolist()

num_bins = 5
min_val = min(grouped_attribute)
max_val = max(grouped_attribute)
width = (max_val - min_val) / num_bins

classes = []
midpoints = []
for i in range(num_bins):
    low = min_val + i * width
    high = min_val + (i + 1) * width
    classes.append((low, high))
    midpoints.append((low + high) / 2)

frequencies = [0] * num_bins
for x in grouped_attribute:
    for i in range(num_bins):
        low = classes[i][0]
        high = classes[i][1]

        if i == num_bins - 1:
            if x >= low and x <= high:
                frequencies[i] += 1
        else:
            if x >= low and x < high:
                frequencies[i] += 1

N = sum(frequencies)

cf = []
cum = 0
for f in frequencies:
    cum += f
    cf.append(cum)

sum_fm = 0
for i in range(num_bins):
    sum_fm += frequencies[i] * midpoints[i]
grp_mean = sum_fm / N

def interpolate_percentile(p):
    target = p * N
    idx = 0

    for i in range(len(cf)):
        if cf[i] >= target:
            idx = i
            break

    L = classes[idx][0]
    if idx > 0:
        prev_cf = cf[idx - 1]
    else:
        prev_cf = 0

    return L + ((target - prev_cf) / frequencies[idx]) * width

grp_median = interpolate_percentile(0.5)
grp_q1 = interpolate_percentile(0.25)
grp_q3 = interpolate_percentile(0.75)
grp_iqr = grp_q3 - grp_q1

max_f = max(frequencies)
modal_idx = frequencies.index(max_f)
L_mode = classes[modal_idx][0]
f1 = frequencies[modal_idx]

if modal_idx > 0:
    f0 = frequencies[modal_idx - 1]
else:
    f0 = 0

if modal_idx < len(frequencies) - 1:
    f2 = frequencies[modal_idx + 1]
else:
    f2 = 0

denom_mode = 2 * f1 - f0 - f2
if denom_mode != 0:
    grp_mode = L_mode + ((f1 - f0) / denom_mode) * width
else:
    grp_mode = midpoints[modal_idx]

var_sum = 0
for i in range(num_bins):
    var_sum += frequencies[i] * ((midpoints[i] - grp_mean) ** 2)
grp_var = var_sum / (N - 1)
grp_std = grp_var ** 0.5

freq_table_data = []
for i in range(num_bins):
    if i < num_bins - 1:
        interval_str = f'[{classes[i][0]:.2f} - {classes[i][1]:.2f})'
    else:
        interval_str = f'[{classes[i][0]:.2f} - {classes[i][1]:.2f}]'

    freq_table_data.append({
        'Class Interval': interval_str,
        'Midpoint (m)': round(midpoints[i], 2),
        'Frequency (f)': frequencies[i],
        'Cumulative Freq (CF)': cf[i],
    })

freq_df = pd.DataFrame(freq_table_data)

grouped_stats_summary = pd.DataFrame([{
    'Grouped Mean': round(grp_mean, 2),
    'Grouped Median': round(grp_median, 2),
    'Grouped Mode': round(grp_mode, 2),
    'Grouped Q1': round(grp_q1, 2),
    'Grouped Q3': round(grp_q3, 2),
    'Grouped IQR': round(grp_iqr, 2),
    'Grouped Variance': round(grp_var, 2),
    'Grouped Std Dev': round(grp_std, 2),
}])

print('ACTIVITY 3: GROUPED DATA FREQUENCY TABLE')
display(freq_df)

print('\nACTIVITY 3: GROUPED DATA STATISTICAL MEASURES')
display(grouped_stats_summary)
```
