```python
import pandas as pd
import matplotlib.pyplot as plt
import math

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("irrigation_prediction.csv")

print("First five rows of the dataset:")
print(df.head())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
def correlation_coefficient(x, y):
    n = len(x)

    sum_x = sum(x)
    sum_y = sum(y)

    sum_xy = sum(a * b for a, b in zip(x, y))
    sum_x2 = sum(a * a for a in x)
    sum_y2 = sum(b * b for b in y)

    numerator = n * sum_xy - sum_x * sum_y

    denominator = math.sqrt(
        (n * sum_x2 - sum_x ** 2) *
        (n * sum_y2 - sum_y ** 2)
    )

    if denominator == 0:
        return float("nan")

    return numerator / denominator

print("Correlation function defined successfully!")
```

```python
x1 = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]

x2 = [1, 2, 3, 4, 5]
y2 = [10, 8, 6, 4, 2]

x3 = [1, 2, 3, 4, 5]
y3 = [2, 5, 3, 1, 4]

print("Positive correlation:", correlation_coefficient(x1, y1))
print("Negative correlation:", correlation_coefficient(x2, y2))
print("No linear correlation:", correlation_coefficient(x3, y3))
```

```python
data = df[["Soil_Moisture", "Rainfall_mm"]].dropna()

x = data["Soil_Moisture"].tolist()
y = data["Rainfall_mm"].tolist()

r = correlation_coefficient(x, y)

print("Number of observations:", len(x))
print("Correlation coefficient:", r)

if not math.isnan(r):
    if r > 0:
        print("Type: Positive correlation")
    elif r < 0:
        print("Type: Negative correlation")
    else:
        print("Type: No linear correlation")
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(x, y, color="blue", alpha=0.2, s=10)

plt.xlabel("Soil Moisture")
plt.ylabel("Rainfall (mm)")
plt.title("Scatter Plot: Soil Moisture vs Rainfall")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
